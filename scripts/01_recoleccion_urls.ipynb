# Recolección de urls de las instituciones universitarias argentinas
---

## Objetivo

* Recolectar en forma automática las **urls de las diferentes instituciones universitarias** integrantes del Sistema Universitario Nacional de la República Argentina mediante la técnica de **web scraping**:
    * UNIVERSIDADES E INSTITUTOS NACIONALES / ESTATALES
    * UNIVERSIDADES E INSTITUTOS PRIVADOS
    * ORGANISMO INTERNACIONAL (FLACSO)
    * INSTITUCIONES PROVINCIALES / JURISDICCIÓN CABA reconocidas
* *Fuente: [Guia de Carreras Universitarias](https://guiadecarreras.siu.edu.ar/instituciones.php)
(guiadecarreras.siu.edu.ar es un sitio web oficial del Gobierno Argentino)*

## Funcionamiento

* **Entrada**: una URL donde se encuentran las instituciones universitarias argentinas (https://guiadecarreras.siu.edu.ar/ciie_ofertas/2.0/listado_instituciones.php)
* **Proceso**: se realiza una solicitud HTTP con la librería requests de Python. Si la conexión es exitosa, se obtiene el contenido HTML de la página mediante la librería BeatifulSoup que permite el parseo del documento facilitando la navegación, búsqueda y extracción de las partes específicas del documento.
* **Salida**: la tabla extraída convertida en un Dataframe, que luego se escribe en el archivo 'universidades-nombre-url.csv'

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from io import StringIO
```

```python
url = 'https://guiadecarreras.siu.edu.ar/ciie_ofertas/2.0/listado_instituciones.php'
```

```python
res = requests.get(url)
soup = BeautifulSoup(res.text, 'html.parser')
tables = soup.find_all('table')
```

```python
df = pd.read_html(StringIO(str(tables)))[0]
```

```python
df[1] = df[1].apply(lambda url: f'https://{url}')
```

```python
df.head()
```

```python
df.to_csv('universidades-nombre-url.csv', sep=';', index=False, header=False)
```
