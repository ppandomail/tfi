# Obtención de status code
---

## Objetivo

* Obtener el status code (códigos numéricos de tres dígitos, ej. 200, 404) de cada una de las URLs de las instituciones universitarias argentinas. Al realizar una solicitud HTTP, el servidor web envía al navegador el código de estado para indicar si la conexión se establece correctamente.

## Funcionamiento

* **Entrada**: cada una de las URLs de las instituciones universitarias argentinas registradas en el archivo 'universidades-nombre-url.csv'
* **Proceso**: para cada URL se realiza una solicitud HTTP con la librería requests de Python.
* **Salida**: el código de estado de la respuesta (ejemplo: 200 si la página responde correctamente)

```python
import requests
import pandas as pd
```

```python
df = pd.read_csv('universidades-nombre-url.csv', sep=';')
```

```python
df.head()
```

```python
# Función para obtener el status_code de cada URL
def get_status(url):
    try:
        response = requests.get(url, verify=False, timeout=5)
        return response.status_code
    except requests.exceptions.RequestException:
        return None  # Si hay error, devuelve None
```

```python
df['status_code'] = df['URL'].apply(lambda url: get_status(url))
```

```python
df
```

```python
df.to_csv('universidades-nombre-url-code.csv', sep=';', index=False)
```
